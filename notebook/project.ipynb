## Feature Selection

The target variable in this project is `Class`, where:

* `0` represents a legitimate transaction.
* `1` represents a fraudulent transaction.

The initial feature set includes the PCA-transformed features `V1`–`V28`, the transaction `Amount`, and the engineered `Hour` feature.

### Selected Features

| Feature    | Decision          | Reason                                                                                                                                |
| ---------- | ----------------- | ------------------------------------------------------------------------------------------------------------------------------------- |
| `V1`–`V28` | Keep              | These are the main PCA-transformed features and may contain useful patterns individually or through interactions with other features. |
| `Amount`   | Keep              | Transaction amount may provide useful information even though its linear correlation with `Class` is very weak.                       |
| `Hour`     | Keep              | Fraud rates vary across different hours, suggesting that time-of-day may contain predictive information.                              |
| `Time`     | Exclude initially | `Hour` was derived from `Time`, so keeping both may introduce redundant temporal information.                                         |
| `Class`    | Target            | This is the variable the models will learn to predict.                                                                                |

### Why not select features only by correlation?

Correlation analysis showed that `V17`, `V14`, `V12`, and `V10` have the strongest linear associations with the target. However, features with weak or near-zero linear correlations are not necessarily useless.

A feature may contribute through:

* Nonlinear relationships with the target.
* Interactions with other features.
* Patterns that cannot be captured by simple pairwise correlation.

Therefore, all `V1`–`V28` features will initially be retained rather than manually removing features based only on their individual correlation with `Class`.

````python
X = df.drop(columns= ['Class', 'Time'])
y = df['Class']
````

````python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=.2, random_state=42, stratify= y)
````

**stratify=y ensures that both sets maintain approximately the same class distribution as the original dataset.**

I cant't do training without make `feature scaling` for more accurate results

````python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
````

## Fit vs Transform

* **`fit`** → The scaler **learns** the necessary statistics (such as mean and standard deviation) from the data.
* **`transform`** → The scaler **uses the learned statistics** to transform the data.
* **`fit_transform`** → Performs both `fit` and `transform` at the same time.

### Important Rule

```text
Training data → fit + transform
Testing data  → transform only
```

We use `fit` only on the training data to avoid **data leakage** from the test set.


````python
from sklearn.linear_model import LogisticRegression

Logistic_Model = LogisticRegression(max_iter=1000)

Logistic_Model.fit(x_train_scaled, y_train)
````

````python
y_pred = Logistic_Model.predict(x_test_scaled)
y_pred
````

````python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
````

- Accuracy  → What percentage of all predictions were correct?

- Precision → Of the transactions the model predicted as Fraud, how many were actually Fraud?

- Recall    → Of all the actual Fraud transactions, how many did the model detect?

- F1 Score  → A balance between Precision and Recall.

**Let's try another model**

````python
from sklearn.ensemble import RandomForestClassifier

Random_forst_model = RandomForestClassifier(
    n_estimators= 100, # 100 Decision Trees
    random_state= 42,
    n_jobs= -1 # Use all he CPU to make it faster 
)

Random_forst_model.fit(x_train, y_train)

y_pred_random_forst = Random_forst_model.predict(x_test)
y_pred_random_forst
````

````python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

print("Accuracy:", accuracy_score(y_test, y_pred_random_forst))
print("Precision:", precision_score(y_test, y_pred_random_forst))
print("Recall:", recall_score(y_test, y_pred_random_forst))
print("F1 Score:", f1_score(y_test, y_pred_random_forst))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_random_forst))
````

Well, Random Forset model impressed me :)
